```python
import schema_ext as schema_ext


def test(
    pipeline, overlay_path: str = "schema_ext_overlay.yaml"
) -> schema_ext.SchemaProvider:
    """
    Convenience factory that assembles the full decorator stack:
        DltSchema → RelationshipDecorator → OverlayDecorator
    """
    return schema_ext.RelationshipDecorator(
        schema_ext.OverlayDecorator(
            schema_ext.DltSchema(pipeline), overlay_path=overlay_path
        )
    )


schema = test(pipeline)

table = schema.get_table("metadata_pe_file")
all_tables = schema.get_all_tables()
```
